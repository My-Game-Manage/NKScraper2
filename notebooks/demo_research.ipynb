```python
# 変更部分を適用するリロードのおまじない
%load_ext autoreload
%autoreload 2

# ロガー準備
import logging
import sys

# 1. rootロガーを取得
logger = logging.getLogger()

# 2. ログレベルを設定 (DEBUG, INFO, WARNING など)
logger.setLevel(logging.INFO)

# 3. Jupyterの標準出力に書き出すハンドラを作成
# すでにハンドラが存在する場合は追加しない（二重出力を防ぐ）
if not logger.handlers:
    handler = logging.StreamHandler(sys.stdout)
    
    # 読みやすいフォーマットを設定
    formatter = logging.Formatter('%(levelname)s: %(name)s: %(message)s')
    handler.setFormatter(formatter)
    
    logger.addHandler(handler)

# テスト: これで engine.py 内のログも表示されるようになります
logging.info("Jupyterのロガー設定が完了しました。")
```

```python
from src.scrape_race_info import ScraperRaceInfo
import json

sc = ScraperRaceInfo()

target_url = 'https://nar.netkeiba.com/race/shutuba.html?race_id=202654092707'

print('データを抽出中...')
extracted_info = sc.get_shutuba_race_info(target_url)

if extracted_info:
    # 抽出した項目データのみをJSONファイルとして保存
    output_filename = 'race_info_summary.json'
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(extracted_info, f, ensure_ascii=False, indent=4)

    print(f'\n【保存完了】以下のレース情報を {output_filename} に保存しました：')
    for key, value in extracted_info.items():
        print(f'- {key}: {value}')
```

```python
from src.scrape_race_info import ScraperRaceInfo
import csv

sc = ScraperRaceInfo()

target_url = 'https://nar.netkeiba.com/race/shutuba.html?race_id=202654092707'

print('データを抽出中...')
extracted_info = sc.get_shutuba_horse_info(target_url)

if extracted_info:
    # 抽出した項目データのみをJSONファイルとして保存
    fieldnames = extracted_info[0].keys()
    output_filename = 'race_info_horselist.csv'
    with open(output_filename, 'w', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(extracted_info)

    print(f'\n【保存完了】以下のレース出馬情報を {output_filename} に保存しました：')
    for l in extracted_info:
        print(f'- {l}')
```
